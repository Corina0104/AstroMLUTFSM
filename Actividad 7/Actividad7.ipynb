En la figura, $\sigma_{NMAD}$ es la [normalized median absolute deviation de los residuos](https://en.wikipedia.org/wiki/Median_absolute_deviation), definida como $\sigma_{NMAD}=1.48\cdot MAD$, donde $MAD = \text{median}(\frac{|\Delta z|}{(1 + zspec)})$, $\Delta z =z_{phot}-z_{spec}$  y $\eta$ es la fracción de outliers definido como los objetos que cumplen $|(\Delta z)| > 0.15\cdot(1+z_{spec})$

Ahora calcule las métricas del paper, fracción de outliers $eta$ y $\sigma_{NMAD}$ y compare con los resultados del paper

```python
z_spec = df[target].values
z_phot = ypred
delta_z = (z_phot - z_spec) / (1 + z_spec)
```

```python
outliers = np.abs(delta_z) > 0.15
eta = (np.sum(outliers) / len(delta_z)) * 100
print(eta)
```

y NMAD

```python
mad = np.median(np.abs(delta_z - np.median(delta_z)))
sigma_nmad = 1.48 * mad
print(sigma_nmad)
```

### Estamos listos para optimizar hiperparámetros...cuáles escogemos?



#### Parámetros de árbol

- El número mínimo de instancias en un nodo hoja

- El número mínimo de instancias requerido en un nodo de división

- La profundidad máxima del árbol

- El criterio elegido para decidir si una división "vale la pena", expresado en términos de ganancia de información

#### Parámetros de randomización



- El número de $k < n$ características que se utilizan para construir los árboles

- El resampling (bootstrap) del conjunto de datos (Verdadero o Falso)

#### Parámetros del bosque

El número de árboles en el bosque (n_estimators) se puede ajustar, con el entendimiento general de que más árboles son mejores, pero en algún punto el rendimiento se estabiliza, por lo que se puede encontrar el equilibrio entre tener más árboles y un menor tiempo de ejecución.

### Conclusion: Cómo se compara nuestro modelo con el resultado del paper? 

Revise la sección 7 del paper. Podemos mejorar nuestro modelo aún más? Implemente un nuevo modelo con la mejora identificada



### Nuestro modelo optimizado muestra un gran avance frente al baseline inicial, aunque todavía presenta una ligera brecha de precisión en comparación con las métricas objetivo reportadas en el paper, para mejorar aún más el desempeño, es posible incorporar índices de color derivados de las magnitudes como variables predictivas adicionales, para capturar de mejor forma la distribución espectral y reducir la dispersión en la estimación de los redshifts. 

```python
sel_features['u-g'] = sel_features['u_apercor'] - sel_features['g_apercor']
sel_features['g-r'] = sel_features['g_apercor'] - sel_features['r_apercor']
sel_features['r-i'] = sel_features['r_apercor'] - sel_features['i_apercor']
sel_features['i-z'] = sel_features['i_apercor'] - sel_features['z_apercor']
sel_features['z-y'] = sel_features['z_apercor'] - sel_features['y_apercor']

scores_new = cross_validate(model, sel_features, sel_target, cv=KFold(n_splits=5, shuffle=True, random_state=10), return_train_score=True)

print(scores_new['test_score'].mean())
```

### Agregar los índices de color elevó el puntaje de prueba del modelo a 0.81, superando la base inicial y final. Esto demuestra que usar las diferencias entre bandas es clave para estimar los redshifts con precisión. Al comparar con el paper, confirmamos que la limpieza de datos y estas nuevas variables nos acercan de buena forma a los resultados del estudio.
